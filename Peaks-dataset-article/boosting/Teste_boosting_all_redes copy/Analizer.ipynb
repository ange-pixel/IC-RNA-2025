```python
import pandas as pd
import numpy as np
from sklearn.metrics import r2_score

def boosting(file_names):
    df_list = [pd.read_excel(arquivo) for arquivo in file_names]
    z_preds = [df['Z_pred'].values.reshape(-1, 1) for df in df_list]
    Z_pred_total = np.hstack(z_preds)
    print(f"Z_pred_total: {Z_pred_total}")
    Z_pred_sum = np.sum(Z_pred_total, axis=1).reshape(-1, 1)

    df = pd.read_excel("1000_model_12_6_0.xlsx")
    Z = df['Z'].values.reshape(-1, 1)

    mse_sup = np.mean((Z - Z_pred_sum) ** 2)
    r2_sup = r2_score(Z, Z_pred_sum)

    print(f"mse: {mse_sup}")
    return mse_sup, r2_sup
file_names= ["1000_model_23_6_4.xlsx", "1000_model_12_6_0.xlsx", "1000_model_13_2_14.xlsx", "1000_model_29_7_11.xlsx", "1000_model_33_7_2.xlsx"]
result= boosting(file_names)
```
