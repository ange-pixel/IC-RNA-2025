```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

df = pd.read_excel('25_model_12_6_0.xlsx')

z_pred = df['Z_pred'].values
z = df['Z'].values
residuos = z-z_pred

# 3️⃣ Gerar o QQ-plot
sm.qqplot(residuos, line='45')

# 4️⃣ Exibir o gráfico
plt.title('QQ-Plot dos Resíduos')
plt.show()
```
