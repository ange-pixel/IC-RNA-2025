```python
import pandas as pd
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import plotly.figure_factory as ff
from scipy.spatial import Delaunay
import plotly.graph_objects as go


def superficie(file_paths):

    # Leitura dos dados e extração de 'Z_pred' reshape em array coluna
    Z_preds = [pd.read_excel(fp)['Z_pred'].values.reshape(-1, 1) for fp in file_paths]
    Z_pred_total = np.hstack(Z_preds)
    Z_media = np.mean(Z_pred_total, axis=1, keepdims=True)
    desvio_padrao = np.std(Z_pred_total, axis=1, keepdims=True)

    df = pd.read_excel("1000_model/1000_model_3_9_0.xlsx")
    Z_mse = df['Z'].values.reshape(-1, 1)

    x = np.linspace(1, 0, 32)
    y = np.linspace(1, 0, 32)
    x, y = np.meshgrid(x, y)
    Z = 1.3356 * (1.5 * (1 - x))+np.exp(2 * x - 1) * np.sin(3 * np.pi * (x - 0.6)**2)+np.exp(3*(y - 0.5)) * np.sin(4 * np.pi * (y - 0.9)**2)
    points = np.column_stack((x.flatten(), y.flatten()))
    tri = Delaunay(points)
            
    fig_original = ff.create_trisurf(x=x.flatten(), y=y.flatten(), z=Z.flatten(), 
                                            simplices=tri.simplices, colormap='Greys',
                                            title='Surface Fit - Original Values')
    fig_predicted = ff.create_trisurf(x=x.flatten(), y=y.flatten(), z= Z_media.flatten(), 
                                            simplices=tri.simplices, colormap='Viridis',
                                            title='Surface Fit - Predicted Values')


    data = [fig_original.data[0], fig_predicted.data[0]]
    fig = go.Figure(data=data)
    fig.update_layout(
    title=' ',
    autosize=False,
    width=800,
    height=800,
    margin=dict(l=65, r=50, b=65, t=90),
    scene=dict(
        aspectmode='manual',
        aspectratio=dict(x=1, y=1, z=1),
        xaxis=dict(
            title=dict(text='x₁', font=dict(family='Times New Roman', size=16)),
            backgroundcolor='white',
            gridcolor='black',
            zerolinecolor='black',
            tickfont=dict(family='Times New Roman', size=14)
        ),
        yaxis=dict(
            title=dict(text='x₂', font=dict(family='Times New Roman', size=16)),
            backgroundcolor='white',
            gridcolor='black',
            zerolinecolor='black',
            tickfont=dict(family='Times New Roman', size=14)
        ),
        zaxis=dict(
            title=dict(text='f(x)', font=dict(family='Times New Roman', size=16)),
            backgroundcolor='white',
            gridcolor='black',
            zerolinecolor='black',
            tickfont=dict(family='Times New Roman', size=14)
        )
    ),
    paper_bgcolor='white',
    plot_bgcolor='white',
    font=dict(family='Times New Roman', size=12)  # fonte global
)


    fig.show()

    mse = np.mean((Z_mse - Z_media) ** 2)

    print(f'MSE: {mse}')

superficie([
    "1000_model/1000_model_11_7_6.xlsx",
    "1000_model/1000_model_11_9_11.xlsx",
    "1000_model/1000_model_11_9_3.xlsx",
    "1000_model/1000_model_14_8_1.xlsx",
    "1000_model/1000_model_14_8_17.xlsx",
    "1000_model/1000_model_14_8_3.xlsx",
    "1000_model/1000_model_3_9_15.xlsx",
    "1000_model/1000_model_7_9_22.xlsx",
    "1000_model/1000_model_9_4_0.xlsx",
    "1000_model/1000_model_9_4_2.xlsx"
])

```
